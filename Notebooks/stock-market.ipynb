```python

# Cell 10: Missing Values Visualization

plt.figure(figsize=(12, 5))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Values in Stock Market Dataset")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python

# Cell 13: Inspect Categorical Columns

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

for col in categorical_columns:
    print("\nColumn:", col)
    print("Unique values:", df[col].nunique(dropna=True))
    print(df[col].value_counts(dropna=False).head(10))
```

```python

# Cell 14: Inspect Numerical Columns

numerical_columns = df.select_dtypes(
    include=np.number
).columns

print("Numerical Columns:")
print(numerical_columns.tolist())

display(df[numerical_columns].describe().T)
```

```python

# Cell 15: Clean Column Names and Text Values

df.columns = df.columns.str.strip()

text_columns = df.select_dtypes(
    include=["object", "category"]
).columns

for col in text_columns:
    df[col] = df[col].apply(
        lambda value: value.strip()
        if isinstance(value, str)
        else value
    )

print("Column names and text whitespace cleaned.")
display(df.head())
```

```python

# Cell 4: ETF Distribution

print(df["ETF"].value_counts(dropna=False))

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="ETF")

plt.title("Distribution of ETF and Non-ETF Securities")
plt.xlabel("ETF (Y = Yes, N = No)")
plt.ylabel("Number of Securities")

plt.tight_layout()
plt.show()
```

### Insight kaise likhna hai?

#### Y = Exchange-Traded Fund.

#### N = Non-ETF security.

#### Graph se dono categories ke records compare honge.

##### Agar ek category ke records zyada hain, toh target imbalance ho sakta hai.

```python

# Cell 5: Listing Exchange Distribution

exchange_counts = df["Listing Exchange"].value_counts()

display(exchange_counts.to_frame("Number of Securities"))

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Listing Exchange",
    order=exchange_counts.index
)

plt.title("Number of Securities by Listing Exchange")
plt.xlabel("Listing Exchange")
plt.ylabel("Number of Securities")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

Insight: Is graph se pata chalega ki dataset mein har listing exchange se kitni securities hain

```python

# Cell 6: Market Category Distribution

market_counts = df["Market Category"].value_counts(
    dropna=False
)

display(market_counts.to_frame("Count"))

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Market Category",
    order=df["Market Category"].value_counts().index
)

plt.title("Distribution of Market Categories")
plt.xlabel("Market Category")
plt.ylabel("Number of Securities")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

Insight: Is graph se different market categories ka distribution samajh aayega. Category codes ka exact financial meaning samajhne ke liye dataset documentation bhi check karni hogi.

```python

# Cell 7: Round Lot Size Distribution

print(df["Round Lot Size"].describe())

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="Round Lot Size",
    bins=20
)

plt.title("Distribution of Round Lot Size")
plt.xlabel("Round Lot Size")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

Insight: Histogram se pata chalega ki kaun-se lot sizes kitni baar appear hote hain aur values ek range mein concentrated hain ya nahi.

```python

# Cell 8: Boxplot for Round Lot Size

plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x="Round Lot Size")

plt.title("Boxplot of Round Lot Size")
plt.xlabel("Round Lot Size")

plt.tight_layout()
plt.show()
```

Insight: Boxplot se distribution aur possible extreme values identify kar sakte hain. Kisi point ka outlier hona zaroori nahi ki woh data error hi ho.

```python

# Cell 9: ETF vs Listing Exchange

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Listing Exchange",
    hue="ETF"
)

plt.title("ETF and Non-ETF Securities by Exchange")
plt.xlabel("Listing Exchange")
plt.ylabel("Number of Securities")
plt.xticks(rotation=30)
plt.legend(title="ETF")

plt.tight_layout()
plt.show()
```

Insight: Is graph se har exchange par ETF aur non-ETF securities ke counts compare honge. Counts exchange ke total size se bhi affect hote hain

```python

# Cell 10: ETF Percentage by Exchange

exchange_percentage = pd.crosstab(
    df["Listing Exchange"],
    df["ETF"],
    normalize="index"
) * 100

display(exchange_percentage.round(2))

exchange_percentage.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)

plt.title("ETF vs Non-ETF Percentage by Exchange")
plt.xlabel("Listing Exchange")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=30)
plt.legend(title="ETF")

plt.tight_layout()
plt.show()
```

Insight: Ye graph har exchange ke andar percentages compare karta hai, isliye exchange sizes alag hone par bhi comparison samajhna easier hota hai.

```python

# Cell 11: Correlation Heatmap

numeric_df = df.select_dtypes(include=np.number)

if numeric_df.shape[1] >= 2:
    correlation = numeric_df.corr()

    plt.figure(figsize=(8, 5))

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()
else:
    print(
        "Not enough numerical columns for a meaningful "
        "correlation heatmap."
    )
```

Insight: Correlation +1 ke kareeb ho toh strong positive linear association, -1 ke kareeb ho toh strong negative linear association aur 0 ke kareeb ho toh weak linear association indicate karta hai. Correlation causation prove nahi karta.

```python
# Correlation Heatmap for Stock Market Dataset

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Convert ETF from categorical to binary
df_corr = df.copy()
df_corr["ETF"] = df_corr["ETF"].map({"N": 0, "Y": 1})

# Select meaningful numeric columns
corr_columns = [
    "Round Lot Size",
    "ETF"
]

corr_matrix = df_corr[corr_columns].corr()

# Plot heatmap
plt.figure(figsize=(7, 5))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Correlation Heatmap of Stock Market Dataset")
plt.tight_layout()
plt.show()
```
